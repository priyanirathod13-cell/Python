```python
#Write a python program to find the volume and area of cylinder
import math
r=float(input("Enter the radius of cylinder : "))
h=float(input("Enter the height of cylinder : "))
print(f"Area Of Cylinder is {2*math.pi*r*(h+r)}")
print(f"Volume Of Cylinder is {math.pi*r*r*h}")
```

```python
# Write a python program to find an integer exponent x such that a^x=n
n = int(input("Enter the number : "))
a = int(input("Enter the base : "))

c = 0
x = a ** c
while x < n:
    c = c + 1
    x = a ** c

# Check if we found the exact match
if n == x:
    print(f"x = {c}")
else:
    print(f"{n} is not a perfect power of {a}")
```
