```python
#pair of function to convert degree to farenheit and farenheit to degree
def DtoF():
    c=int(input("Enter celsius : "))
    print(f"Temperature in farenheit is {(c*9/5)+32}")
DtoF()
def FToD():
    f=int(input("Enter farenheit : "))
    print(f"Temperature in celsius is {(f-32)*5/9}")
FToD()
```

```python
#function to check number is in given range or not
def check(number,start,end):
    if start<=number<=end:
        return True
    else:
        return False
number=float(input("Enter Any Number : "))
start=float(input("Enter Starting Point : "))
end=float(input("Enter Ending Point : "))
if check(number,start,end):
    print("Number is in Range")
else:
    print("Number is not in range")
```

```python
#factorial of a given number
def fact(n):
    result=1
    if n<0:
        return None
    for i in range(1,n+1):
        result = result*i
    return result
n=int(input("Enter non-negative integer : "))
answer=fact(n)
if answer is None:
    print("Factorial not defined")
else:
    print("Factorial = ",answer)
```

```python
#Fibonacci series
def fibbo(n):
    a=0
    b=1
    for i in range(n):
        print(a," ")
        c=a+b
        a=b
        b=c
n=int(input("Enter no. of terms : "))
if n>0:
    fibbo(n)
else:
    print("Enter Positive number")
```

```python
#MAX OF TWO
def findMax(a,b):
    if a>b:
        return a
    return b
x=float(input("Enter First Number : "))
y=float(input("Enter Second Number : "))
print(f"Max of two is {findMax(x,y)}")
```

```python
#odd even checker
def odd_even(num):
    if num%2==0:
        return "Even"
    return "Odd"
a=int(input("Enter Any Number : "))
b=int(input("Enter Any Number : "))
print(a,"is",odd_even(a))
print(b,"is",odd_even(b))
```

```python
#Armstrong
def armstrong(n):
    if n<0:
        return False
    temp=n
    digit=0
    if temp==0:
        digit=1
    else:
        while temp>0:
            digit=digit+1
            temp=temp//10
    temp=n
    total=0
    while temp>0:
        d=temp%10
        total=total+(d**digit)
        temp=temp//10
    return total==n
n=int(int(input("Enter Any Number : ")))
if armstrong(n):
    print("Number is Armstrong")
else:
    print("Number is not Armstrong")
```

```python
#calculator using function
def Add(a,b):
    return a+b
def Sub(a,b):
    return a-b
def Mul(a,b):
    return a*b
def Div(a,b):
    if a==b:
        return None
    return a//b
a=float(input("Enter Any Number : "))
o=input("Enter Any operator : ")
b=float(input("Enter Any Number : "))
if o=="+":
    print(f"Result : {Add(a,b)}")
elif o=="-":
    print(f"Result : {Sub(a,b)}")
elif o=="*":
    print(f"Result : {Mul(a,b)}")
elif o=="/":
    result=Div(a,b)
    if result is None:
        print("cannot divide by zero")
    else:
        print(f"Result : {result}")
else:
    print("Invalid")
```
