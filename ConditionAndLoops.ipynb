```python
#write a python program to compute the product of odd numbers in a given number
a = int(input("Enter any number : "))
n = 1
odd_no = False

while a > 0:
    r = a % 10
    if r % 2 != 0:
        n = n * r
        odd_no = True
    a = a // 10
if odd_no:
    print(f"The product is : {n}")
else:
    print("No Odd Number found")
```

```python
#Write a program to check if a number is prime or not
n=int(input("Enter any number : "))
if n<2:
    print("Number is not prime")
else:
    prime = True
    for i in range(2,n):
        if n%i==0:
            prime=False
            break
    if prime:
        print("Prime Number")
    else:
        print("Not Prime Number")
```

```python
#Write a program to print prime numbers between given interval entered by user 
s=int(input("Enter starting point : "))
e=int(input("Enter ending point : "))
for i in range(s,e+1):
    if i>=2:
        prime=True
        for j in range(2,i):
            if i%j==0:
                prime=False
                break
    if prime:
        print(i,end=" ")
```

```python
#Disarium number
for i in range(1, 101):
    x = i
    x1 = i
    digit = 0
    
    # 1. Count the total number of digits
    while x1 > 0:
        x1 = x1 // 10
        digit = digit + 1
        
    # 2. Reset r to 0 for the current number 'i'
    r = 0
    temp_i = i  # Use a temporary variable so we don't destroy the loop counter 'i'
    
    # 3. Process digits from right to left
    while temp_i > 0:
        d = temp_i % 10
        r = r + (d ** digit)  # Rightmost digit gets the highest remaining power
        digit = digit - 1     # Move the power down for the next digit to the left
        temp_i = temp_i // 10
        
    # 4. Check if it matches the original number
    if r == x:
        print(f"{x} is disarium")
```
